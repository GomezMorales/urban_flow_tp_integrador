# Ejercicio 03

> Puntos: 1

Diseñar el modelo lógico según el archivo data/processed/speeding_fines_image.csv de las siguientes entidades:

- Vehiculo

- Multa

- Radar

- Evidencia

Relaciones que deben ser consideradas a la hora del diseño:

- Un vehículo puede tener muchas multas.

- Una multa puede tener una evidencia (opcional).

- Un radar puede generar muchas multas.

El modelo lógico no está relacionado con la base de datos.

````python
#@title Modelo lógico: clases de dominio

@dataclass
class Vehiculo:
  """Representa un vehículo registrado en el sistema de multas.

  Attributes:
      patente: Matrícula única del vehículo.
  """

  patente: str


@dataclass
class Radar:
  """Representa un radar urbano capaz de registrar infracciones.

  Attributes:
      radar_id:  Identificador único del radar.
      ubicacion: Descripción de la ubicación física del radar.
  """

  radar_id: str
  ubicacion: str


@dataclass
class Evidencia:
  """Registra la evidencia visual opcional asociada a una multa.

  Attributes:
      archivo_imagen:  Nombre del archivo de imagen capturado.
      patente_imagen:  Patente leída automáticamente desde la imagen.
      ratio:           Nivel de confianza del reconocimiento óptico.
  """

  archivo_imagen: str
  patente_imagen: str
  ratio: float


@dataclass
class Multa:
  """Representa una infracción por exceso de velocidad.

  Relaciones:
    - Cada Multa pertenece a exactamente un Vehiculo.
    - Cada Multa es generada por exactamente un Radar.
    - Cada Multa puede tener a lo sumo una Evidencia (opcional).

  Attributes:
      multa_id:             Identificador único de la multa.
      fecha:                Fecha en que se registró la infracción.
      hora:                 Hora en que se registró la infracción.
      velocidad_registrada: Velocidad detectada por el radar (km/h).
      velocidad_maxima:     Límite de velocidad permitido en el tramo
                            (km/h).
      estado_multa:         Estado administrativo (IMPAGA, PAGADA,
                            etc.).
      vehiculo:             Vehículo infractor asociado a la multa.
      radar:                Radar que generó la multa.
      evidencia:            Evidencia visual adjunta (puede ser None).
  """

  multa_id: str
  fecha: str
  hora: str
  velocidad_registrada: float
  velocidad_maxima: float
  estado_multa: str
  vehiculo: Vehiculo
  radar: Radar
  evidencia: Optional[Evidencia] = None
````

# Ejercicio 04

> Puntos: 1

Completar la función `procesar_fila_csv` que recibe como parámetro un `dict` y debe cargar los datos de una fila del archivo `data/processed/speeding_fines_image.csv` y mapearla según las clases definidas en el punto anterior.

````python
#@title Función procesar_fila_csv
def procesar_fila_csv(row: dict) -> Multa:
  """Mapea una fila del CSV a una instancia del modelo lógico.

  Construye las instancias de Vehiculo, Radar y Evidencia (cuando
  corresponde) a partir del diccionario recibido y las vincula dentro
  de un objeto Multa.

  Args:
      row: Diccionario con las claves del archivo
           data/processed/speeding_fines_image.csv.

  Returns:
      Multa: Instancia completamente inicializada con sus relaciones.
  """
  vehiculo = Vehiculo(patente=row["patente"])

  radar = Radar(
    radar_id=row["radar_id"],
    ubicacion=row["ubicacion"],
  )

  imagen = row.get("imagen")
  evidencia: Optional[Evidencia] = None
  if imagen and not pd.isna(imagen):
    evidencia = Evidencia(
      archivo_imagen=str(imagen),
      patente_imagen=str(row.get("patente_imagen", "")),
      ratio=float(row.get("ratio", 0.0)),
    )

  return Multa(
    multa_id=row["multa_id"],
    fecha=str(row["fecha"]),
    hora=str(row["hora"]),
    velocidad_registrada=float(row["velocidad_registrada"]),
    velocidad_maxima=float(row["velocidad_maxima"]),
    estado_multa=row["estado_multa"],
    vehiculo=vehiculo,
    radar=radar,
    evidencia=evidencia,
  )
````

````python
#@title Validación de la función

row_ejemplo = {
  'multa_id': 'M-999',
  'patente': 'AA123BB',
  'fecha': '2023-10-25',
  'hora': '14:30:00',
  'velocidad_registrada': 130,
  'velocidad_maxima': 110,
  'ubicacion': 'Au. 25 de Mayo Km 4',
  'radar_id': 'R-01',
  'estado_multa': 'IMPAGA',
  'imagen': 'foto_123.jpg',
  'patente_imagen': 'AA12388',
  'ratio': 0.85
}

multa_objeto = procesar_fila_csv(row_ejemplo)

print(f"Multa creada para el auto: {multa_objeto.vehiculo.patente}")
print(f"Generada por el radar en: {multa_objeto.radar.ubicacion}")
if multa_objeto.evidencia:
  print(f"Evidencia adjunta: {multa_objeto.evidencia.archivo_imagen}")
````

````python
class Base(DeclarativeBase):
    pass

class Vehiculo(Base):
    __tablename__ = 'vehiculos'
    patente = Column(String, primary_key=True)
    multas = relationship("Multa", back_populates="vehiculo")

    def __repr__(self):
        return f"<Vehiculo(patente='{self.patente}')>"

class Radar(Base):
    __tablename__ = 'radares'
    radar_id = Column(String, primary_key=True)
    ubicacion = Column(String)
    multas = relationship("Multa", back_populates="radar")

    def __repr__(self):
        return f"<Radar(radar_id='{self.radar_id}', ubicacion='{self.ubicacion}')>"

class Evidencia(Base):
    __tablename__ = 'evidencias'
    evidencia_id = Column(Integer, primary_key=True, autoincrement=True)
    archivo_imagen = Column(String)
    patente_imagen = Column(String)
    ratio = Column(Float)
    multa_id = Column(String, ForeignKey('multas.multa_id'), unique=True, nullable=False)
    multa = relationship("Multa", back_populates="evidencia")

    def __repr__(self):
        return f"<Evidencia(evidencia_id={self.evidencia_id}, archivo_imagen='{self.archivo_imagen}', multa_id='{self.multa_id}')>"

class Multa(Base):
    __tablename__ = 'multas'
    multa_id = Column(String, primary_key=True)
    fecha = Column(Date)
    hora = Column(Time)
    velocidad_registrada = Column(Float)
    velocidad_maxima = Column(Float)
    estado_multa = Column(String)

    patente = Column(String, ForeignKey('vehiculos.patente'), nullable=False)
    radar_id = Column(String, ForeignKey('radares.radar_id'), nullable=True)

    vehiculo = relationship("Vehiculo", back_populates="multas")
    radar = relationship("Radar", back_populates="multas")
    evidencia = relationship("Evidencia", uselist=False, back_populates="multa", cascade="all, delete-orphan")

    def __repr__(self):
        return f"<Multa(multa_id='{self.multa_id}', patente='{self.patente}', radar_id='{self.radar_id}', fecha='{self.fecha}')>"
````

##Funciones para crear y poblar de la Base de Datos

````python
DATABASE_URL = "sqlite:///transito.db"
engine = create_engine(DATABASE_URL, echo=False)
csv_path = "urban_flow/data/processed/speeding_fines_image.csv"

def migrar_datos():
    # Eliminar BD previa para evitar conflictos de claves primarias
    if os.path.exists("transito.db"):
        os.remove("transito.db")

    print("Creando tablas automáticamente...")
    Base.metadata.create_all(engine)

    if not os.path.exists(csv_path):
        print(f"Error: No se encontró el archivo CSV en: {csv_path}")
        return

    df = pd.read_csv(csv_path)
    print(f"Archivo CSV leído con éxito. Filas detectadas: {len(df)}")

    # Tratamiento inteligente de fechas y horas
    df['fecha_limpia'] = pd.to_datetime(df['fecha']).dt.date
    df['hora_limpia'] = pd.to_datetime(df['hora'], format='mixed').dt.time

    with Session(engine) as session:
        try:
            vehiculos_registrados = set()
            radares_registrados = set()

            for idx, row in df.iterrows():
                # Saltear registros rotos
                if pd.isna(row['patente']) or pd.isna(row['multa_id']):
                    continue

                patente = str(row['patente']).strip().upper()
                multa_id = str(row['multa_id']).strip().upper()

                if patente == "NAN" or multa_id == "NAN":
                    continue

                # --- Mapear Radar ---
                radar_id = None
                if pd.notna(row.get('radar_id')) and str(row['radar_id']).strip().lower() != "nan":
                    radar_id = str(row['radar_id']).strip().upper()

                    if radar_id not in radares_registrados:
                        ubicacion_texto = str(row['ubicacion']).strip().upper() if pd.notna(row.get('ubicacion')) else "SIN UBICACIÓN"
                        radar = Radar(radar_id=radar_id, ubicacion=ubicacion_texto)
                        session.add(radar)
                        radares_registrados.add(radar_id)

                # --- Mapear Vehículo ---
                if patente not in vehiculos_registrados:
                    vehiculo = Vehiculo(patente=patente)
                    session.add(vehiculo)
                    vehiculos_registrados.add(patente)

                # --- Mapear Multa ---
                multa = Multa(
                    multa_id=multa_id,
                    fecha=row['fecha_limpia'],
                    hora=row['hora_limpia'],
                    velocidad_registrada=float(row['velocidad_registrada']),
                    velocidad_maxima=float(row['velocidad_maxima']),
                    estado_multa=str(row['estado_multa']).strip().upper() if pd.notna(row['estado_multa']) else "PENDIENTE",
                    patente=patente,
                    radar_id=radar_id
                )
                session.add(multa)

                # --- Mapear Evidencia ---
                if pd.notna(row.get('imagen')) and str(row['imagen']).strip().lower() != "nan":
                    archivo_img = str(row['imagen']).strip()
                    if archivo_img != "":
                        evidencia = Evidencia(
                            archivo_imagen=archivo_img,
                            patente_imagen=str(row['patente_imagen']).strip().upper() if pd.notna(row['patente_imagen']) else patente,
                            ratio=float(row['ratio']) if pd.notna(row['ratio']) else 0.0,
                            multa_id=multa_id
                        )
                        session.add(evidencia)

            session.commit()
            print("¡Migración completada con éxito!")

        except Exception as e:
            session.rollback()
            print(f"Error crítico en migración: {e}")
            raise e

def validar_registros():
    print("\n--- VALIDACIÓN DE REGISTROS INSERTADOS ---")

    if os.path.exists(csv_path):
        df = pd.read_csv(csv_path)
        total_validas_csv = df['multa_id'].dropna().count()
        evidencias_csv = df['imagen'].dropna().count()
    else:
        total_validas_csv = 0
        evidencias_csv = 0

    with Session(engine) as session:
        cant_vehiculos = session.scalar(select(func.count(Vehiculo.patente)))
        cant_radares = session.scalar(select(func.count(Radar.radar_id)))
        cant_multas = session.scalar(select(func.count(Multa.multa_id)))
        cant_evidencias = session.scalar(select(func.count(Evidencia.evidencia_id)))

        print(f"Multas esperadas (CSV)            : {total_validas_csv}")
        print(f"Imágenes/Evidencias esperadas     : {evidencias_csv}")
        print("--------------------------")
        print(f"Vehículos creados (BD)            : {cant_vehiculos}")
        print(f"Radares creados (BD)              : {cant_radares}")
        print(f"Multas reales insertadas (BD)     : {cant_multas}")
        print(f"Evidencias reales vinculadas (BD) : {cant_evidencias}")

        if cant_multas == total_validas_csv:
            print("El total de registros en la Base de Datos es exactamente igual al archivo CSV original.")
        else:
            print("Verificación terminada (se aplicaron filtros de limpieza).")
````
