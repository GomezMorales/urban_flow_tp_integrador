group_images = {
  'plates': [{

- Desarrollar una función para mostrar 8 imágenes de forma aleatoria en forma
  de tabla de 2 columnas y 4 filas. Debe mostrar 4 imágenes de cada grupo. Es
  una función reutilizable. En este punto se deben mostrar las imágenes
  originales. Cada imagen debe tener un ancho de 10 pulgadas y un alto de 8
  pulgadas, se debe mostrar el nombre de la imagen.

````python
os.makedirs(PATH_INTERIM, exist_ok=True)
JSON_PATH = os.path.join(PATH_INTERIM, "group_images.json")

group_images = {
    'plates': [],
    'completes': []
}

total_area = 0
image_count = 0
lista_imagenes_disponibles = []

for root, dirs, files in os.walk(PATH_RAW_IMGS):
  for file in files:
    if file.lower().endswith(('.png', '.jpg', '.jpeg')):
      full_path = os.path.join(root, file)

      size_kb = os.path.getsize(full_path) / 1024

      lista_imagenes_disponibles.append({
        'nombre': file,
        'tamano_kb': round(size_kb, 2)
      })

      img = cv2.imread(full_path)
      if img is not None:
        height, width = img.shape[:2]
        total_area += (width * height)
        image_count += 1

area_promedio = total_area / image_count if image_count > 0 else 0

for root, dirs, files in os.walk(PATH_RAW_IMGS):
  for file in files:
    if file.lower().endswith(('.png', '.jpg', '.jpeg', '.jp2')):
      full_path = os.path.join(root, file)

      img = cv2.imread(full_path)
      if img is not None:
        height, width = img.shape[:2]
        area_actual = width * height

        if area_actual < area_promedio:
          grupo = 'plates'
        else:
          grupo = 'completes'

          img_metadata = {
            'filename': file,
            'width': width,
            'height': height,
            'area': area_actual,
            'path': os.path.abspath(full_path),
            'patent': ''
          }

          group_images[grupo].append(img_metadata)

with open(JSON_PATH, 'w', encoding='utf-8') as f:
  json.dump(group_images, f, indent=4, ensure_ascii=False)

print(f"El grupo 'plates' contiene {len(group_images['plates'])} imágenes.")
print(f"El grupo 'completes' contiene {len(group_images['completes'])} imágenes.")
print("---------------------------------------")
print("--- LISTADO DE IMÁGENES DISPONIBLES ---")
for img in lista_imagenes_disponibles:
  print(f"Nombre: {img['nombre']} | Tamaño: {img['tamano_kb']:.2f} KB")
````

````python

def mostrar_imagenes_aleatorias(diccionario_grupos: dict):
  """
  Selecciona 4 imágenes aleatorias de 'plates' y 4 de 'completes'
  y las muestra en una cuadrícula de 2 columnas por 4 filas.
  Cada imagen individual se visualiza a 10 pulgadas de ancho por 8 de alto.
  """
  if len(diccionario_grupos['plates']) < 4 or len(diccionario_grupos['completes']) < 4:
    print("Error: Se requieren al menos 4 imágenes en cada grupo para realizar el muestreo.")
    return

  muestra_plates = random.sample(diccionario_grupos['plates'], 4)
  muestra_completes = random.sample(diccionario_grupos['completes'], 4)

  lista_final = muestra_plates + muestra_completes

  fig, axes = plt.subplots(nrows=4, ncols=2, figsize=(20, 32))
  axes = axes.flatten()

  for i, info in enumerate(lista_final):
    img_bgr = cv2.imread(info['path'])

    if img_bgr is not None:
      img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
      axes[i].imshow(img_rgb)
    else:
      axes[i].text(0.5, 0.5, "Error al cargar", ha='center', va='center')

      axes[i].set_title(info['filename'], fontsize=14, pad=10)

      axes[i].axis('off')

  plt.tight_layout()
  plt.show()
````

En este punto realizaremos un procesamiento de las imágenes, en cada punto se debe llamar a la función que muestra las imágenes trabajadas en el punto anterior.

- Conversión a escala de grises de las imágenes originales y almacenarlas en `urban_flow/data/interim/imgs/03_01_gray_scale/plates` y `urban_flow/data/interim/imgs/03_01_gray_scale/completes`, según la clasificación del ejercicio 2.   

- Realizar un suavizado de las imágenes en escala de grises y almacenarlas en `urban_flow/data/interim/imgs/03_02_blur` según la clasificación del ejercicio 2.

- Realizar una detección de bordes sobre las imágenes suavizadas para intentar detectar las patentes y almacenarlas en `urban_flow/data/interim/imgs/03_03_canny` según la clasificación del ejercicio 2.



````python
# Título: Ejercicio 03 - Paso 1: Conversión a escala de grises
PATH_GRAY = "urban_flow/data/interim/imgs/03_01_gray_scale"

def procesar_grises(group_images: dict) -> dict:
  """Convierte las imágenes originales a escala de grises y las guarda."""
  group_gray = {'plates': [], 'completes': []}

  for grupo in ['plates', 'completes']:
    carpeta_salida = os.path.join(PATH_GRAY, grupo)
    os.makedirs(carpeta_salida, exist_ok=True)

    for img_info in group_images[grupo]:
      img_bgr = cv2.imread(img_info['path'])

      if img_bgr is None:
        print(f"No se pudo leer: {img_info['path']}")
        continue

      img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
      ruta_salida = os.path.join(carpeta_salida, img_info['filename'])
      cv2.imwrite(ruta_salida, img_gray)
      group_gray[grupo].append(
        {**img_info, 'path': os.path.abspath(ruta_salida)}
      )
  return group_gray

group_gray = procesar_grises(group_images)
mostrar_imagenes_aleatorias(group_gray)
````

````python
# Título: Ejercicio 03 - Paso 2: Suavizado de imágenes en escala de grises
PATH_BLUR = "urban_flow/data/interim/imgs/03_02_blur"

def procesar_blur(group_gray: dict) -> dict:
  """Aplica suavizado gaussiano a las imágenes en escala de grises."""
  group_blur = {'plates': [], 'completes': []}

  for grupo in ['plates', 'completes']:
    carpeta_salida = os.path.join(PATH_BLUR, grupo)
    os.makedirs(carpeta_salida, exist_ok=True)

    for img_info in group_gray[grupo]:
      img_gray = cv2.imread(img_info['path'], cv2.IMREAD_GRAYSCALE)

      if img_gray is None:
        print(f"No se pudo leer: {img_info['path']}")
        continue

      img_blur = cv2.GaussianBlur(img_gray, (5, 5), 0)
      ruta_salida = os.path.join(carpeta_salida, img_info['filename'])
      cv2.imwrite(ruta_salida, img_blur)
      group_blur[grupo].append(
        {**img_info, 'path': os.path.abspath(ruta_salida)}
      )
  return group_blur

group_blur = procesar_blur(group_gray)
mostrar_imagenes_aleatorias(group_blur)
````

````python
# Título: Ejercicio 03 - Paso 3: Detección de bordes con Canny
PATH_CANNY = "urban_flow/data/interim/imgs/03_03_canny"

def procesar_canny(group_blur: dict) -> dict:
  """Aplica detección de bordes Canny a las imágenes suavizadas."""
  group_canny = {'plates': [], 'completes': []}

  for grupo in ['plates', 'completes']:
    carpeta_salida = os.path.join(PATH_CANNY, grupo)
    os.makedirs(carpeta_salida, exist_ok=True)

    for img_info in group_blur[grupo]:
      img_blur = cv2.imread(img_info['path'], cv2.IMREAD_GRAYSCALE)

      if img_blur is None:
        print(f"No se pudo leer: {img_info['path']}")
        continue

      img_canny = cv2.Canny(img_blur, 100, 200)
      ruta_salida = os.path.join(carpeta_salida, img_info['filename'])
      cv2.imwrite(ruta_salida, img_canny)
      group_canny[grupo].append(
        {**img_info, 'path': os.path.abspath(ruta_salida)}
      )
  return group_canny

group_canny = procesar_canny(group_blur)
mostrar_imagenes_aleatorias(group_canny)
````

````python
# Título: Ejercicio 04 - Funciones modulares del pipeline OCR


def listar_imagenes(carpeta: str) -> List[Path]:
  """
  Lista de forma eficiente las imágenes de una carpeta.

  Usa pathlib para iterar el directorio y filtra solo
  archivos con extensiones de imagen válidas.

  Args:
    carpeta: Ruta a la carpeta de imágenes a escanear.

  Returns:
    Lista ordenada de objetos Path a imágenes válidas.
  """
  ruta = Path(carpeta)
  extensiones = {".jpg", ".jpeg", ".png"}
  return sorted([
    p for p in ruta.iterdir()
    if p.is_file() and p.suffix.lower() in extensiones
  ])


def inicializar_reader_ocr() -> easyocr.Reader:
  """
  Inicializa el Reader de EasyOCR para inglés y español.

  Se configura sin GPU para mayor compatibilidad en Colab.
  La inicialización descarga los modelos la primera vez.

  Returns:
    Instancia de easyocr.Reader lista para usar.
  """
  return easyocr.Reader(['en', 'es'], gpu=False)


def extraer_texto_imagen(
  reader: easyocr.Reader,
  ruta: Path,
) -> str:
  """
  Extrae el texto de una imagen de patente usando EasyOCR.

  Captura cualquier excepción para que el pipeline no se
  detenga si una imagen está corrupta o no tiene texto.

  Args:
    reader: Lector EasyOCR ya inicializado.
    ruta:   Ruta Path a la imagen a procesar.

  Returns:
    Texto detectado concatenado, o cadena vacía si falla.
  """
  try:
    resultados = reader.readtext(str(ruta))
    if not resultados:
      return ""
    return " ".join(item[1] for item in resultados)
  except Exception as e:
    print(f"  [ERROR] {ruta.name}: {e}")
    return ""


def limpiar_texto_ocr(texto: str) -> str:
  """
  Normaliza el string extraído por OCR.

  Convierte a mayúsculas y elimina todo carácter que no
  sea letra (A-Z) ni dígito (0-9), ya que el ruido de la
  imagen Canny puede introducir símbolos extraños.

  Args:
    texto: String crudo devuelto por EasyOCR.

  Returns:
    String limpio: solo A-Z y 0-9 en mayúsculas.
  """
  texto = texto.upper()
  return re.sub(r'[^A-Z0-9]', '', texto)


def calcular_similitud(cadena_a: str, cadena_b: str) -> float:
  """
  Calcula el porcentaje de similitud entre dos strings.

  Usa difflib.SequenceMatcher, que encuentra los bloques
  de caracteres coincidentes más largos y aplica:

      ratio = 2 * M / T

  donde M = cantidad de caracteres en bloques coincidentes
        T = total de caracteres en ambas cadenas juntas.

  Ejemplos prácticos:
    - "ABC123" vs "ABC123" → ratio = 1.0  (idénticos)
    - "ABC123" vs "AB0123" → ratio ≈ 0.83 (1 char diff)
    - "ABC123" vs "XYZ999" → ratio ≈ 0.0  (sin match)

  El parámetro `isjunk=None` desactiva la heurística de
  autobasura, logrando mayor precisión en strings cortos
  como patentes donde cada carácter tiene peso relevante.

  Args:
    cadena_a: Primer string a comparar.
    cadena_b: Segundo string a comparar.

  Returns:
    Float en [0.0, 1.0]: 1.0 = idénticos, 0.0 = sin match.
  """
  if not cadena_a or not cadena_b:
    return 0.0
  matcher = difflib.SequenceMatcher(
    None, cadena_a, cadena_b
  )
  return matcher.ratio()


def buscar_mejor_coincidencia(
  texto_ocr: str,
  df: pd.DataFrame,
  col_patente: str = COL_PATENTE,
  umbral: float = UMBRAL_SIMILITUD,
) -> Dict[str, Any]:
  """
  Busca en el DataFrame la patente con mayor similitud
  al texto extraído por OCR.

  Itera todas las filas del DataFrame, normaliza cada
  patente (misma limpieza que el OCR) y calcula similitud.
  La fila con mayor similitud que supere el umbral
  determina si hay match exitoso.

  Args:
    texto_ocr:   Texto ya limpiado por limpiar_texto_ocr.
    df:          DataFrame de infracciones con patentes.
    col_patente: Nombre de la columna de patentes en df.
    umbral:      Similitud mínima para match (default 0.80).

  Returns:
    Dict con claves:
      match (bool): True si similitud >= umbral.
      similitud (float): Mejor similitud encontrada.
      patente_esperada (str|None): Patente del DataFrame.
      multa_id (int|None): ID de la multa coincidente.
  """
  mejor_similitud: float = 0.0
  mejor_fila: Optional[pd.Series] = None

  for _, fila in df.iterrows():
    patente_norm = re.sub(
      r'[^A-Z0-9]', '', str(fila[col_patente]).upper()
    )
    sim = calcular_similitud(texto_ocr, patente_norm)
    if sim > mejor_similitud:
      mejor_similitud = sim
      mejor_fila = fila

  return {
    "match": mejor_similitud >= umbral,
    "similitud": round(mejor_similitud, 4),
    "patente_esperada": (
      mejor_fila[col_patente]
      if mejor_fila is not None else None
    ),
    "multa_id": (
      int(mejor_fila["multa_id"])
      if mejor_fila is not None else None
    ),
  }


def procesar_ocr_pipeline(
  reader: easyocr.Reader,
  imagenes: List[Path],
  df: pd.DataFrame,
  umbral: float = UMBRAL_SIMILITUD,
) -> List[Dict[str, Any]]:
  """
  Ejecuta el pipeline OCR completo sobre las imágenes.

  Para cada imagen aplica en secuencia:
    1. Extracción de texto con EasyOCR.
    2. Limpieza del texto (mayúsculas, solo alfanumérico).
    3. Búsqueda del mejor match en el DataFrame.

  Args:
    reader:   Reader EasyOCR ya inicializado.
    imagenes: Lista de rutas Path a imágenes de patentes.
    df:       DataFrame con patentes reales esperadas.
    umbral:   Umbral mínimo de similitud para match.

  Returns:
    Lista de dicts con los resultados por imagen.
  """
  resultados: List[Dict[str, Any]] = []

  for ruta in imagenes:
    texto_crudo = extraer_texto_imagen(reader, ruta)
    texto_limpio = limpiar_texto_ocr(texto_crudo)
    resultado = buscar_mejor_coincidencia(
      texto_limpio, df, umbral=umbral
    )
    estado = "✓ MATCH" if resultado["match"] else "✗ ---  "
    print(
      f"  {ruta.name[:38]:<38} "
      f"OCR: {texto_limpio:<10} | {estado}"
    )
    resultados.append({
      "imagen": str(ruta),
      "texto_ocr_crudo": texto_crudo,
      "texto_ocr_limpio": texto_limpio,
      **resultado,
    })

  return resultados
````

````python
# Título: Ejercicio 04 - Ejecución del pipeline OCR

df_interim: pd.DataFrame = pd.read_csv(PATH_INTERIM_CSV)

imagenes_canny: List[Path] = listar_imagenes(PATH_CANNY_PLATES)
print(
  f"Imágenes detectadas en carpeta Canny plates: "
  f"{len(imagenes_canny)}\n"
)

reader_ocr: easyocr.Reader = inicializar_reader_ocr()

resultados_ocr: List[Dict[str, Any]] = procesar_ocr_pipeline(
  reader_ocr,
  imagenes_canny,
  df_interim,
  umbral=UMBRAL_SIMILITUD,
)
print(f"\nPipeline finalizado. Imágenes procesadas: {len(resultados_ocr)}")
````

````python
# Título: Ejercicio 04 - Integración al DataFrame y resumen

df_ocr_resultados: pd.DataFrame = pd.DataFrame(resultados_ocr)

df_matches: pd.DataFrame = (
  df_ocr_resultados[df_ocr_resultados["match"]]
  [["multa_id", "imagen", "texto_ocr_limpio", "similitud"]]
  .dropna(subset=["multa_id"])
  .astype({"multa_id": int})
  .drop_duplicates(subset=["multa_id"], keep="first")
)

df_procesado_final: pd.DataFrame = df_interim.merge(
  df_matches.rename(columns={
    "texto_ocr_limpio": "patente_imagen",
    "similitud": "ratio",
  }),
  on="multa_id",
  how="left",
)

try:
  ocr_por_imagen: Dict[str, Dict] = {
    Path(r["imagen"]).name: r for r in resultados_ocr
  }
  for img_info in group_canny["plates"]:
    nombre = img_info["filename"]
    if nombre in ocr_por_imagen:
      img_info["patent"] = (
        ocr_por_imagen[nombre]["texto_ocr_limpio"]
      )
  print("group_canny actualizado con patentes OCR.\n")
except NameError:
  print(
    "[AVISO] group_canny no está en memoria. "
    "Ejecutar celdas del Ejercicio 03 primero.\n"
  )

PATH_PROCESSED_CSV: str = (
  "urban_flow/data/processed/speeding_fines_image.csv"
)
df_procesado_final.to_csv(PATH_PROCESSED_CSV, index=False)

total_imgs: int = len(df_ocr_resultados)
total_match: int = int(df_ocr_resultados["match"].sum())
pct_exito: float = (
  total_match / total_imgs * 100 if total_imgs > 0 else 0.0
)

print("=" * 52)
print("  RESUMEN EJERCICIO 04 — OCR EasyOCR + Match 80%")
print("=" * 52)
print(f"  Imágenes procesadas    : {total_imgs}")
print(f"  Matches exitosos (≥80%): {total_match}")
print(f"  Porcentaje de éxito    : {pct_exito:.1f}%")
print("=" * 52)

print("\n--- Vista previa del DataFrame con OCR ---")
cols_vista: List[str] = [
  "multa_id", "patente", "imagen", "patente_imagen", "ratio"
]
print(
  df_procesado_final[cols_vista]
  .dropna(subset=["patente_imagen"])
  .head(10)
  .to_string(index=False)
)
````
