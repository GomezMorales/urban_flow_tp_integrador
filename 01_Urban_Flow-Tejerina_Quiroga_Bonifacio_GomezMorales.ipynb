# Ejercicio 03

> Puntos: 1

Diseñar el modelo lógico según el archivo data/processed/speeding_fines_image.csv de las siguientes entidades:

- Vehiculo

- Multa

- Radar

- Evidencia

Relaciones que deben ser consideradas a la hora del diseño:

- Un vehículo puede tener muchas multas.

- Una multa puede tener una evidencia (opcional).

- Un radar puede generar muchas multas.

El modelo lógico no está relacionado con la base de datos.

````python
#@title Modelo lógico: clases de dominio

@dataclass
class Vehiculo:
  """Representa un vehículo registrado en el sistema de multas.

  Attributes:
      patente: Matrícula única del vehículo.
  """

  patente: str


@dataclass
class Radar:
  """Representa un radar urbano capaz de registrar infracciones.

  Attributes:
      radar_id:  Identificador único del radar.
      ubicacion: Descripción de la ubicación física del radar.
  """

  radar_id: str
  ubicacion: str


@dataclass
class Evidencia:
  """Registra la evidencia visual opcional asociada a una multa.

  Attributes:
      archivo_imagen:  Nombre del archivo de imagen capturado.
      patente_imagen:  Patente leída automáticamente desde la imagen.
      ratio:           Nivel de confianza del reconocimiento óptico.
  """

  archivo_imagen: str
  patente_imagen: str
  ratio: float


@dataclass
class Multa:
  """Representa una infracción por exceso de velocidad.

  Relaciones:
    - Cada Multa pertenece a exactamente un Vehiculo.
    - Cada Multa es generada por exactamente un Radar.
    - Cada Multa puede tener a lo sumo una Evidencia (opcional).

  Attributes:
      multa_id:             Identificador único de la multa.
      fecha:                Fecha en que se registró la infracción.
      hora:                 Hora en que se registró la infracción.
      velocidad_registrada: Velocidad detectada por el radar (km/h).
      velocidad_maxima:     Límite de velocidad permitido en el tramo
                            (km/h).
      estado_multa:         Estado administrativo (IMPAGA, PAGADA,
                            etc.).
      vehiculo:             Vehículo infractor asociado a la multa.
      radar:                Radar que generó la multa.
      evidencia:            Evidencia visual adjunta (puede ser None).
  """

  multa_id: str
  fecha: str
  hora: str
  velocidad_registrada: float
  velocidad_maxima: float
  estado_multa: str
  vehiculo: Vehiculo
  radar: Radar
  evidencia: Optional[Evidencia] = None
````

# Ejercicio 04

> Puntos: 1

Completar la función `procesar_fila_csv` que recibe como parámetro un `dict` y debe cargar los datos de una fila del archivo `data/processed/speeding_fines_image.csv` y mapearla según las clases definidas en el punto anterior.

````python
#@title Función procesar_fila_csv
def procesar_fila_csv(row: dict) -> Multa:
  """Mapea una fila del CSV a una instancia del modelo lógico.

  Construye las instancias de Vehiculo, Radar y Evidencia (cuando
  corresponde) a partir del diccionario recibido y las vincula dentro
  de un objeto Multa.

  Args:
      row: Diccionario con las claves del archivo
           data/processed/speeding_fines_image.csv.

  Returns:
      Multa: Instancia completamente inicializada con sus relaciones.
  """
  vehiculo = Vehiculo(patente=row["patente"])

  radar = Radar(
    radar_id=row["radar_id"],
    ubicacion=row["ubicacion"],
  )

  imagen = row.get("imagen")
  evidencia: Optional[Evidencia] = None
  if imagen and not pd.isna(imagen):
    evidencia = Evidencia(
      archivo_imagen=str(imagen),
      patente_imagen=str(row.get("patente_imagen", "")),
      ratio=float(row.get("ratio", 0.0)),
    )

  return Multa(
    multa_id=row["multa_id"],
    fecha=str(row["fecha"]),
    hora=str(row["hora"]),
    velocidad_registrada=float(row["velocidad_registrada"]),
    velocidad_maxima=float(row["velocidad_maxima"]),
    estado_multa=row["estado_multa"],
    vehiculo=vehiculo,
    radar=radar,
    evidencia=evidencia,
  )
````

````python
#@title Validación de la función

row_ejemplo = {
  'multa_id': 'M-999',
  'patente': 'AA123BB',
  'fecha': '2023-10-25',
  'hora': '14:30:00',
  'velocidad_registrada': 130,
  'velocidad_maxima': 110,
  'ubicacion': 'Au. 25 de Mayo Km 4',
  'radar_id': 'R-01',
  'estado_multa': 'IMPAGA',
  'imagen': 'foto_123.jpg',
  'patente_imagen': 'AA12388',
  'ratio': 0.85
}

multa_objeto = procesar_fila_csv(row_ejemplo)

print(f"Multa creada para el auto: {multa_objeto.vehiculo.patente}")
print(f"Generada por el radar en: {multa_objeto.radar.ubicacion}")
if multa_objeto.evidencia:
  print(f"Evidencia adjunta: {multa_objeto.evidencia.archivo_imagen}")
````

````python
class Base(DeclarativeBase):
    pass

class Vehiculo(Base):
    __tablename__ = 'vehiculos'
    patente = Column(String, primary_key=True)
    multas = relationship("Multa", back_populates="vehiculo")

    def __repr__(self):
        return f"<Vehiculo(patente='{self.patente}')>"

class Radar(Base):
    __tablename__ = 'radares'
    radar_id = Column(String, primary_key=True)
    ubicacion = Column(String)
    multas = relationship("Multa", back_populates="radar")

    def __repr__(self):
        return f"<Radar(radar_id='{self.radar_id}', ubicacion='{self.ubicacion}')>"

class Evidencia(Base):
    __tablename__ = 'evidencias'
    evidencia_id = Column(Integer, primary_key=True, autoincrement=True)
    archivo_imagen = Column(String)
    patente_imagen = Column(String)
    ratio = Column(Float)
    multa_id = Column(String, ForeignKey('multas.multa_id'), unique=True, nullable=False)
    multa = relationship("Multa", back_populates="evidencia")

    def __repr__(self):
        return f"<Evidencia(evidencia_id={self.evidencia_id}, archivo_imagen='{self.archivo_imagen}', multa_id='{self.multa_id}')>"

class Multa(Base):
    __tablename__ = 'multas'
    multa_id = Column(String, primary_key=True)
    fecha = Column(Date)
    hora = Column(Time)
    velocidad_registrada = Column(Float)
    velocidad_maxima = Column(Float)
    estado_multa = Column(String)

    patente = Column(String, ForeignKey('vehiculos.patente'), nullable=False)
    radar_id = Column(String, ForeignKey('radares.radar_id'), nullable=True)

    vehiculo = relationship("Vehiculo", back_populates="multas")
    radar = relationship("Radar", back_populates="multas")
    evidencia = relationship("Evidencia", uselist=False, back_populates="multa", cascade="all, delete-orphan")

    def __repr__(self):
        return f"<Multa(multa_id='{self.multa_id}', patente='{self.patente}', radar_id='{self.radar_id}', fecha='{self.fecha}')>"
````

##Funciones para crear y poblar de la Base de Datos

````python
DATABASE_URL = "sqlite:///transito.db"
engine = create_engine(DATABASE_URL, echo=False)
csv_path = "urban_flow/data/processed/speeding_fines_image.csv"

def migrar_datos():
    # Eliminar BD previa para evitar conflictos de claves primarias
    if os.path.exists("transito.db"):
        os.remove("transito.db")

    print("Creando tablas automáticamente...")
    Base.metadata.create_all(engine)

    if not os.path.exists(csv_path):
        print(f"Error: No se encontró el archivo CSV en: {csv_path}")
        return

    df = pd.read_csv(csv_path)
    print(f"Archivo CSV leído con éxito. Filas detectadas: {len(df)}")

    # Tratamiento inteligente de fechas y horas
    df['fecha_limpia'] = pd.to_datetime(df['fecha']).dt.date
    df['hora_limpia'] = pd.to_datetime(df['hora'], format='mixed').dt.time

    with Session(engine) as session:
        try:
            vehiculos_registrados = set()
            radares_registrados = set()

            for idx, row in df.iterrows():
                # Saltear registros rotos
                if pd.isna(row['patente']) or pd.isna(row['multa_id']):
                    continue

                patente = str(row['patente']).strip().upper()
                multa_id = str(row['multa_id']).strip().upper()

                if patente == "NAN" or multa_id == "NAN":
                    continue

                # --- Mapear Radar ---
                radar_id = None
                if pd.notna(row.get('radar_id')) and str(row['radar_id']).strip().lower() != "nan":
                    radar_id = str(row['radar_id']).strip().upper()

                    if radar_id not in radares_registrados:
                        ubicacion_texto = str(row['ubicacion']).strip().upper() if pd.notna(row.get('ubicacion')) else "SIN UBICACIÓN"
                        radar = Radar(radar_id=radar_id, ubicacion=ubicacion_texto)
                        session.add(radar)
                        radares_registrados.add(radar_id)

                # --- Mapear Vehículo ---
                if patente not in vehiculos_registrados:
                    vehiculo = Vehiculo(patente=patente)
                    session.add(vehiculo)
                    vehiculos_registrados.add(patente)

                # --- Mapear Multa ---
                multa = Multa(
                    multa_id=multa_id,
                    fecha=row['fecha_limpia'],
                    hora=row['hora_limpia'],
                    velocidad_registrada=float(row['velocidad_registrada']),
                    velocidad_maxima=float(row['velocidad_maxima']),
                    estado_multa=str(row['estado_multa']).strip().upper() if pd.notna(row['estado_multa']) else "PENDIENTE",
                    patente=patente,
                    radar_id=radar_id
                )
                session.add(multa)

                # --- Mapear Evidencia ---
                if pd.notna(row.get('imagen')) and str(row['imagen']).strip().lower() != "nan":
                    archivo_img = str(row['imagen']).strip()
                    if archivo_img != "":
                        evidencia = Evidencia(
                            archivo_imagen=archivo_img,
                            patente_imagen=str(row['patente_imagen']).strip().upper() if pd.notna(row['patente_imagen']) else patente,
                            ratio=float(row['ratio']) if pd.notna(row['ratio']) else 0.0,
                            multa_id=multa_id
                        )
                        session.add(evidencia)

            session.commit()
            print("¡Migración completada con éxito!")

        except Exception as e:
            session.rollback()
            print(f"Error crítico en migración: {e}")
            raise e

def validar_registros():
    print("\n--- VALIDACIÓN DE REGISTROS INSERTADOS ---")

    if os.path.exists(csv_path):
        df = pd.read_csv(csv_path)
        total_validas_csv = df['multa_id'].dropna().count()
        evidencias_csv = df['imagen'].dropna().count()
    else:
        total_validas_csv = 0
        evidencias_csv = 0

    with Session(engine) as session:
        cant_vehiculos = session.scalar(select(func.count(Vehiculo.patente)))
        cant_radares = session.scalar(select(func.count(Radar.radar_id)))
        cant_multas = session.scalar(select(func.count(Multa.multa_id)))
        cant_evidencias = session.scalar(select(func.count(Evidencia.evidencia_id)))

        print(f"Multas esperadas (CSV)            : {total_validas_csv}")
        print(f"Imágenes/Evidencias esperadas     : {evidencias_csv}")
        print("--------------------------")
        print(f"Vehículos creados (BD)            : {cant_vehiculos}")
        print(f"Radares creados (BD)              : {cant_radares}")
        print(f"Multas reales insertadas (BD)     : {cant_multas}")
        print(f"Evidencias reales vinculadas (BD) : {cant_evidencias}")

        if cant_multas == total_validas_csv:
            print("El total de registros en la Base de Datos es exactamente igual al archivo CSV original.")
        else:
            print("Verificación terminada (se aplicaron filtros de limpieza).")
````

## Top 10 patentes con más multas

````python
def top_patentes_con_multas(
  session: Session,
  limite: int = 10,
) -> List[Tuple[str, int]]:
  """Obtiene las patentes con mayor cantidad de multas.

  Args:
    session: Sesión activa de SQLAlchemy.
    limite: Cantidad máxima de patentes a retornar.

  Returns:
    Lista de tuplas (patente, cantidad) en orden descendente.
  """
  consulta = (
    select(Multa.patente, func.count(Multa.multa_id).label("cantidad"))
    .group_by(Multa.patente)
    .order_by(desc("cantidad"))
    .limit(limite)
  )
  return session.execute(consulta).all()


def imprimir_top_patentes(resultados: List[Tuple[str, int]]) -> None:
  """Imprime el ranking de patentes con el formato solicitado.

  Args:
    resultados: Lista de tuplas (patente, cantidad).
  """
  print("--- Top 10 patentes con mayor cantidad de multas ---")
  print(f"{'Patente':<10} | {'Cantidad':<10}")
  print(f"{'-' * 10} | {'-' * 10}")
  for patente, cantidad in resultados:
    print(f"{patente:<10} | {cantidad:<10}")


with Session(engine) as session:
  top_patentes = top_patentes_con_multas(session)
  imprimir_top_patentes(top_patentes)
````

## Top 10 multas sin evidencia

````python
def multas_sin_evidencia(
  session: Session,
  limite: int = 10,
) -> List[Multa]:
  """Obtiene las multas que no tienen evidencia asociada.

  Se ordena por fecha de forma ascendente para priorizar las
  multas más antiguas que aún no cuentan con evidencia visual,
  lo que aporta más valor analítico que ordenar por el
  identificador de la multa. Se observa que algunos registros
  conservan la fecha por defecto (1932-01-01) proveniente de la
  limpieza de datos, por lo que aparecen primeros en el listado.

  Args:
    session: Sesión activa de SQLAlchemy.
    limite: Cantidad máxima de multas a retornar.

  Returns:
    Lista de objetos Multa sin evidencia, orden ascendente
    por fecha.
  """
  consulta = (
    select(Multa)
    .outerjoin(Evidencia)
    .where(Evidencia.evidencia_id.is_(None))
    .order_by(Multa.fecha.asc())
    .limit(limite)
  )
  return session.scalars(consulta).all()


def contar_multas_sin_evidencia(session: Session) -> int:
  """Cuenta el total de multas sin evidencia asociada.

  Args:
    session: Sesión activa de SQLAlchemy.

  Returns:
    Cantidad total de multas sin evidencia.
  """
  consulta = (
    select(func.count(Multa.multa_id))
    .outerjoin(Evidencia)
    .where(Evidencia.evidencia_id.is_(None))
  )
  return session.scalar(consulta)


def imprimir_multas_sin_evidencia(
  multas: List[Multa],
  total: int,
) -> None:
  """Imprime el ranking de multas sin evidencia.

  Args:
    multas: Lista de multas a mostrar.
    total: Total de multas sin evidencia.
  """
  print(f"--- Total multas sin evidencia ({total}) ---")
  print("--- Top 10 ---")
  print(f"{'Multa':<10} | {'Fecha':<10}")
  print(f"{'-' * 10} | {'-' * 10}")
  for multa in multas:
    print(f"{multa.multa_id:<10} | {str(multa.fecha):<10}")


with Session(engine) as session:
  total_sin_evidencia = contar_multas_sin_evidencia(session)
  multas = multas_sin_evidencia(session)
  imprimir_multas_sin_evidencia(multas, total_sin_evidencia)
````

## Radares más activos

````python
def radares_mas_activos(
  session: Session,
  limite: int = 10,
) -> List[Tuple[str, str, int]]:
  """Obtiene los radares con mayor cantidad de infracciones.

  Args:
    session: Sesión activa de SQLAlchemy.
    limite: Cantidad máxima de radares a retornar.

  Returns:
    Lista de tuplas (radar_id, ubicacion, cantidad) descendente.
  """
  consulta = (
    select(
      Radar.radar_id,
      Radar.ubicacion,
      func.count(Multa.multa_id).label("cantidad"),
    )
    .join(Multa)
    .group_by(Radar.radar_id, Radar.ubicacion)
    .order_by(desc("cantidad"))
    .limit(limite)
  )
  return session.execute(consulta).all()


def imprimir_radares_mas_activos(
  resultados: List[Tuple[str, str, int]],
) -> None:
  """Imprime el ranking de radares con el formato solicitado.

  Args:
    resultados: Lista de tuplas (radar_id, ubicacion, cantidad).
  """
  print("--- Radares más activos ---")
  print(f"{'Radar':<10} | {'Ubicación':<20} | {'Cantidad':<10}")
  print(f"{'-' * 10} | {'-' * 20} | {'-' * 10}")
  for radar_id, ubicacion, cantidad in resultados:
    print(f"{radar_id:<10} | {ubicacion:<20} | {cantidad:<10}")


with Session(engine) as session:
  radares = radares_mas_activos(session)
  imprimir_radares_mas_activos(radares)
````

## Top 10 patentes reincidentes en un período

````python
def top_reincidentes_periodo(
  session: Session,
  desde: date,
  hasta: date,
  limite: int = 10,
) -> List[Tuple[str, int]]:
  """Obtiene las patentes más reincidentes en un período dado.

  Args:
    session: Sesión activa de SQLAlchemy.
    desde: Fecha inicial del período (inclusive).
    hasta: Fecha final del período (inclusive).
    limite: Cantidad máxima de patentes a retornar.

  Returns:
    Lista de tuplas (patente, cantidad) en orden descendente.
  """
  consulta = (
    select(Multa.patente, func.count(Multa.multa_id).label("cantidad"))
    .where(Multa.fecha.between(desde, hasta))
    .group_by(Multa.patente)
    .order_by(desc("cantidad"))
    .limit(limite)
  )
  return session.execute(consulta).all()


def imprimir_reincidentes(
  resultados: List[Tuple[str, int]],
  desde: date,
  hasta: date,
) -> None:
  """Imprime el ranking de reincidentes con el formato solicitado.

  Args:
    resultados: Lista de tuplas (patente, cantidad).
    desde: Fecha inicial del período.
    hasta: Fecha final del período.
  """
  print(f"--- Top 10 Reincidentes ({desde} a {hasta}) ---")
  print(f"{'Patente':<10} | {'Cantidad':<10}")
  print(f"{'-' * 10} | {'-' * 10}")
  for patente, cantidad in resultados:
    print(f"{patente:<10} | {cantidad:<10}")


with Session(engine) as session:
  desde = date(2020, 1, 1)
  hasta = date(2020, 12, 31)
  reincidentes = top_reincidentes_periodo(session, desde, hasta)
  imprimir_reincidentes(reincidentes, desde, hasta)
````

## Porcentaje de multas confirmadas visualmente

````python
def estadisticas_evidencia(session: Session) -> Tuple[int, int, float]:
  """Calcula el porcentaje de multas con evidencia visual.

  Args:
    session: Sesión activa de SQLAlchemy.

  Returns:
    Tupla (total_multas, con_foto, porcentaje).
  """
  total_multas = session.scalar(select(func.count(Multa.multa_id)))
  con_foto = session.scalar(select(func.count(Evidencia.evidencia_id)))
  porcentaje = (con_foto / total_multas * 100) if total_multas else 0.0
  return total_multas, con_foto, porcentaje


def imprimir_estadisticas_evidencia(
  total_multas: int,
  con_foto: int,
  porcentaje: float,
) -> None:
  """Imprime las estadísticas de evidencia con el formato pedido.

  Args:
    total_multas: Cantidad total de multas.
    con_foto: Cantidad de multas con evidencia.
    porcentaje: Porcentaje confirmado visualmente.
  """
  print("--- Estadísticas de Evidencia ---")
  print(f"Total Multas: {total_multas}")
  print(f"Con Foto: {con_foto}")
  print(f"Porcentaje Confirmado: {porcentaje:.2f}%")


with Session(engine) as session:
  total, con_foto, porcentaje = estadisticas_evidencia(session)
  imprimir_estadisticas_evidencia(total, con_foto, porcentaje)
````

## Cargar modelo

````python
# ViT-B-32 es una arquitectura liviana, rápida y eficiente de OpenCLIP.
NOMBRE_MODELO: str = "ViT-B-32"
PESOS_MODELO: str = "laion2b_s34b_b79k"

dispositivo: str = "cuda" if torch.cuda.is_available() else "cpu"

modelo_clip, _, preprocess = open_clip.create_model_and_transforms(
  NOMBRE_MODELO,
  pretrained=PESOS_MODELO,
)
modelo_clip = modelo_clip.to(dispositivo)
modelo_clip.eval()

print(f"Modelo {NOMBRE_MODELO} cargado en {dispositivo}.")
````

## Funcion para generar el embedding

````python
def generar_embedding(ruta_imagen: str) -> List[float]:
  """Genera el vector de características de una imagen con OpenCLIP.

  Args:
    ruta_imagen: Ruta al archivo de imagen.

  Returns:
    Lista de floats que representa el embedding normalizado.
  """
  imagen = Image.open(ruta_imagen).convert("RGB")
  tensor = preprocess(imagen).unsqueeze(0).to(dispositivo)

  with torch.no_grad():
    caracteristicas = modelo_clip.encode_image(tensor)
    caracteristicas /= caracteristicas.norm(dim=-1, keepdim=True)

  return caracteristicas.squeeze(0).cpu().tolist()
````

## Crear y poblar base de datos vectorial

````python
DIR_IMAGENES: str = "urban_flow/data/raw/imgs/urban_flow_plates"
NOMBRE_COLECCION: str = "patente_vectorial"


def obtener_pares_evidencia(
  session: Session,
) -> List[Tuple[str, str]]:
  """Obtiene los pares (patente, archivo_imagen) desde la BD.

  Args:
    session: Sesión activa de SQLAlchemy.

  Returns:
    Lista de tuplas (patente, archivo_imagen) con evidencia.
  """
  consulta = (
    select(Multa.patente, Evidencia.archivo_imagen)
    .join(Evidencia, Multa.multa_id == Evidencia.multa_id)
  )
  return session.execute(consulta).all()


def poblar_base_vectorial(
  coleccion: chromadb.Collection,
  pares: List[Tuple[str, str]],
) -> int:
  """Genera embeddings y los carga en la colección vectorial.

  Args:
    coleccion: Colección de ChromaDB a poblar.
    pares: Lista de tuplas (patente, archivo_imagen).

  Returns:
    Cantidad de vectores insertados correctamente.
  """
  insertados = 0
  for patente, archivo_imagen in pares:
    ruta = archivo_imagen
    if not os.path.exists(ruta):
      continue

    vector = generar_embedding(ruta)
    coleccion.add(
      ids=[patente],
      embeddings=[vector],
      metadatas=[{"archivo_imagen": archivo_imagen}],
    )
    insertados += 1

  return insertados


cliente_chroma = chromadb.Client()
coleccion_patentes = cliente_chroma.get_or_create_collection(
  name=NOMBRE_COLECCION,
)

with Session(engine) as session:
  pares_evidencia = obtener_pares_evidencia(session)

cantidad = poblar_base_vectorial(coleccion_patentes, pares_evidencia)
print(f"Vectores insertados en '{NOMBRE_COLECCION}': {cantidad}")
````

## Funciones para buscar patente y vehículo

````python
def buscar_vehiculo_relacional(
  session: Session,
  patente: str,
) -> Optional[Dict[str, Any]]:
  """Recupera los datos de un vehículo y sus multas desde la BD.

  Args:
    session: Sesión activa de SQLAlchemy.
    patente: Patente del vehículo a buscar.

  Returns:
    Diccionario con los datos del vehículo, o None si no existe.
  """
  vehiculo = session.get(Vehiculo, patente)
  if vehiculo is None:
    return None

  return {
    "patente": vehiculo.patente,
    "cantidad_multas": len(vehiculo.multas),
    "multas": [
      {
        "multa_id": m.multa_id,
        "fecha": str(m.fecha),
        "velocidad_registrada": m.velocidad_registrada,
        "estado_multa": m.estado_multa,
      }
      for m in vehiculo.multas
    ],
  }


def buscar_patente_imagen(imagen: str) -> Optional[Dict[str, Any]]:
  """Identifica un vehículo a partir de una imagen de patente.

  Genera el embedding de la imagen, busca la patente más
  similar en la base vectorial y recupera los datos del
  vehículo desde la base relacional.

  Args:
    imagen: Ruta a la imagen de la patente a identificar.

  Returns:
    Diccionario con los datos del vehículo y la distancia de
    similitud, o None si no se encuentra coincidencia.
  """
  vector = generar_embedding(imagen)

  resultado = coleccion_patentes.query(
    query_embeddings=[vector],
    n_results=1,
  )

  if not resultado["ids"] or not resultado["ids"][0]:
    return None

  patente = resultado["ids"][0][0]
  distancia = resultado["distances"][0][0]

  with Session(engine) as session:
    datos = buscar_vehiculo_relacional(session, patente)

  if datos is None:
    return None

  datos["distancia"] = distancia
  return datos
````

````python
%%writefile -a urban_flow/data/Readme.md

# Conclusiones — Sprint 3

## Resumen del trabajo desarrollado

En este sprint se consolidó el pipeline de datos de Urban Flow,
pasando de un manejo basado en archivos a una arquitectura con
versionado, persistencia relacional y búsqueda vectorial.

El versionado se separó en dos planos: Git para el código y los
punteros livianos, y DVC para los binarios pesados (imágenes), con
un remote local como respaldo. Esta separación evita inflar el
repositorio con archivos grandes y permite reconstruir cualquier
estado de los datos.

Sobre el modelo de dominio se definió un modelo relacional con
SQLAlchemy (Vehiculo, Radar, Multa y Evidencia) y se pobló la base
`transito` a partir del CSV procesado. Luego se implementaron
consultas analíticas (reincidencia, radares más activos, multas sin
evidencia, confirmación visual) que demuestran el valor de tener los
datos normalizados en un motor relacional.

Finalmente se incorporó una capa de búsqueda por similitud visual:
los embeddings de las patentes se generaron con OpenCLIP y se
almacenaron en ChromaDB, vinculando cada vector a su vehículo. La
función `buscar_patente_imagen` combina ambos mundos: recupera la
patente más cercana en la base vectorial y completa los datos desde
la base relacional.

## Análisis crítico

El sistema funciona de extremo a extremo, pero el análisis dejó al
descubierto una limitación que se origina en etapas previas. El OCR
del Sprint 2 se ejecutó sobre las imágenes de patentes y solo
el 23% superó el umbral de match del 80%. El reconocimiento de
caracteres tuvo una tasa de acierto baja, con lecturas que agregaban
o perdían caracteres respecto de la patente real. Como consecuencia,
la base de datos quedó con apenas 20 evidencias, y la base vectorial
heredó ese volumen reducido.

Esa decisión temprana se propagó hasta la búsqueda vectorial. Al
validar `buscar_patente_imagen` con las imágenes originales a color,
las distancias resultaron altas (cercanas a 1.0), porque la colección
fue poblada con imágenes Canny y los embeddings de una imagen a color
y su versión Canny no son comparables. Al repetir la validación con
imágenes Canny —el mismo dominio que se almacenó— las distancias
bajaron a 0.0 con la patente correcta, confirmando que la lógica de
búsqueda es correcta y que la limitación está en el tipo de imagen.

## Mejoras propuestas

El punto más claro que dejaron las pruebas es la importancia de la
consistencia: la base vectorial debe poblarse y consultarse siempre
con el mismo tipo de imagen. En la validación, las consultas con
imágenes del mismo dominio que el almacenado dieron coincidencia
exacta, mientras que mezclar dominios distintos produjo distancias
altas. Garantizar esa coherencia entre lo que se almacena y lo que
se consulta es la condición mínima para que la búsqueda por
similitud sea confiable. Una forma de lograrlo, sin alterar lo ya
almacenado, sería aplicar a la imagen de consulta el mismo
preprocesamiento que atravesaron las imágenes de la base (escala de
grises, blur y Canny) antes de generar su embedding, de modo que
ambas queden en el mismo dominio.

Más allá de eso, sería valioso revisar la estrategia de
preprocesamiento de imágenes (escala de grises, blur, Canny) usada en
etapas previas. La baja tasa de reconocimiento del OCR sugiere que
vale la pena experimentar con distintas combinaciones de filtros y
comparar sus resultados, en lugar de asumir que la cadena actual es la
óptima. No se midió cuál preprocesamiento rinde mejor ni para el OCR
ni para los embeddings, de modo que esto queda como una línea de
trabajo a evaluar empíricamente en futuras iteraciones.
````
