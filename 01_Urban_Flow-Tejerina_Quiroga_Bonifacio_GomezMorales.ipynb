### Normalización

#### Normalización campo "patente"

````python
#Para tratar las patentes, primero creo una columna llamada "patente_clean",
# normalizo a mayúsculas todas las patentes
df_original['patente_clean'] = df_original['patente'].str.upper()

#Realizo los siguientes replaces:
# Quito los espacios (usando regex para ser más robusto con cualquier tipo 
# de espacio)
# Quito los guiones
# Convierto el valor interpretable nulo "INVALID_PLATE" por NaN

#Por último, fuerzo que el campo sea considerado un string nullable, preservando pd.NA.
df_original['patente_clean'] = df_original['patente_clean'].str\
.strip()\
.replace(r'\s+', '', regex=True)\
.str.replace('-', '', regex=False)\
.replace('INVALID_PLATE', pd.NA)\
.astype(pd.StringDtype())
````

#### Normalización campo "fecha"

````python
# Limpieza del campo fecha
# 1. Convierte 'fecha' a minúsculas, elimina espacios en blanco y reemplaza
# '/' con '-' para un formato consistente.
# 2. Asigna el resultado a 'fecha_clean' para usarla como columna auxiliar.
df_original['fecha_clean'] = df_original['fecha'].str.lower().str.strip().str.replace('/', '-', regex=False)
# 3. Reemplaza la cadena específica 'invalid_date' con pd.NaT para que pd.to_datetime pueda manejarla como un valor faltante.
df_original['fecha_clean'] = df_original['fecha_clean'].replace('invalid_date', pd.NaT)

# Intenta parsear fechas con dayfirst=True (ej., 'dd-mm-yyyy')
parsed_dayfirst = pd.to_datetime(df_original['fecha_clean'], errors='coerce', dayfirst=True)

# Intenta parsear fechas con formato año-primero (ej., 'yyyy-mm-dd')
parsed_yearfirst = pd.to_datetime(df_original['fecha_clean'], errors='coerce', format='%Y-%m-%d')

# Combina las series parseadas: Prioriza day-first, luego year-first.
# Si parsed_dayfirst devuelve NaT, intenta parsed_yearfirst y asigna el resultado a 'fecha_clean'.
df_original['fecha_clean'] = parsed_dayfirst.fillna(parsed_yearfirst)

# Finalmente, reemplaza cualquier valor NaT restante (aquellos que fallaron 
# en todos los intentos de parseo) con un objeto datetime para '1932-01-01'
df_original['fecha_clean'] = df_original['fecha_clean'].fillna(pd.to_datetime('1932-01-01'))
````

#### Normalización campo "hora"

````python
#Normalizo a mayúsculas AM/PM para hacer la conversión y normalizo el valor
# inválido que utiliza el df ("invalid_hour") para que sea mas sencillo
# cambiarlo a NaN.
df_original['hora_clean'] = df_original['hora'].str.lower().replace('invalid_hour', pd.NA)

#Convierto primero las horas en formato 24hs
parsed_24h = pd.to_datetime(df_original['hora_clean'], format='%H:%M', errors='coerce').dt.time

#Convierto las horas que este en formato AM/PM
parsed_12h = pd.to_datetime(df_original['hora_clean'].str.upper(), format='%I:%M %p', errors='coerce').dt.time

#En este paso creo una nueva columna "hora_clean" y ahi guardo las horas
# casteadas, para ello combino parsed_24 contra parsed_12 donde si el 
# primero tiene un valor válido coloca ese, si encuentra un NaT (Not a Time)
# lo rellena con el valor equivalente en parsed_12h

df_original['hora_clean'] = parsed_24h.fillna(parsed_12h)

#Paso final, aplico un lambda a la columna "hora_clean" para que aplique la
# conversión al formato deseado (HH:MM), los valores inválidos son
# transformados a '00:00'
df_original['hora_clean'] = df_original['hora_clean'].apply(lambda x: x.strftime('%H:%M') if pd.notna(x) else '00:00')
````

#### Normalización campo "ubicación"

````python
#Normalizo primero a mayúsculas las direcciones, luego quito espacios que
# puedan existir a izquierda y derecha de la cadena

#Normalizo mediante regex la variación de tipeo en AV
# El primer regex "^\s*AV\." va a tomar todos los espacios (si existiesen)
#  junto a "AV." y los va a reemplazar por "AV "
# El segundo regex "\s+" va a tomar la multiple existencia de espacios y los 
# va a reemplazar por un único espacio



df_original['ubicacion_clean'] = df_original['ubicacion'].str\
  .upper()\
  .str.strip()\
  .replace(r'^\s*AV\.','AV ',regex=True)\
  .replace(r'\s+', ' ', regex=True)\
  .astype(str)
````

### Asignar las columnas normalizadas a las originales para que quede listo el dataset

````python
df_original['fecha'] = df_original['fecha_clean']
df_original['hora'] = df_original['hora_clean']
df_original['ubicacion'] = df_original['ubicacion_clean']
df_original['patente'] = df_original['patente_clean']

#Elimino las columnas *_clean que sirvieron para la normalización de la
# información
df_original = df_original.drop(columns=['fecha_clean', 'hora_clean', 'ubicacion_clean','patente_clean'])
````

### Detección y eliminación de Outliers

````python
def remove_outliers_iqr(df: pd.DataFrame, columns: List[str]) -> pd.DataFrame:
  """
  Elimina outliers de un DataFrame usando el método IQR para las columnas especificadas.

  Args:
    df (pd.DataFrame): El DataFrame de entrada.
    columns (List[str]): Lista de nombres de columnas numéricas a procesar.

  Returns:
    pd.DataFrame: El DataFrame sin outliers.
    int: Cantidad de filas eliminadas.
  """
  initial_rows = df.shape[0]
  df_cleaned_outliers = df.copy()
  # Usaremos un conjunto para asegurar que solo se elimine una vez por cada fila que sea outlier en MÚLTIPLES columnas
  outlier_indices = set()
  cols_with_outliers_detected = [] # Renombrado para evitar confusión con el parámetro 'columns'

  for col in columns:
    if col in df_cleaned_outliers.columns and pd.api.types.is_numeric_dtype(df_cleaned_outliers[col]):
      Q1 = df_cleaned_outliers[col].quantile(0.25)
      Q3 = df_cleaned_outliers[col].quantile(0.75)
      IQR = Q3 - Q1
      lower_bound = Q1 - 1.5 * IQR
      upper_bound = Q3 + 1.5 * IQR

      # Identificar outliers por índice
      col_outliers = df_cleaned_outliers[(df_cleaned_outliers[col] < lower_bound) | (df_cleaned_outliers[col] > upper_bound)].index

      if not col_outliers.empty:
        cols_with_outliers_detected.append(col)
        outlier_indices.update(col_outliers.tolist())

  if outlier_indices:
    df_cleaned_outliers = df_cleaned_outliers.drop(index=list(outlier_indices))

  final_rows = df_cleaned_outliers.shape[0]
  removed_count = initial_rows - final_rows

  print(f"Se eliminaron {removed_count} filas.")
  print("Las columnas observadas son:")
  print("\n".join(f"- {column}" for column in columns))

  return df_cleaned_outliers


#Aplicación de la función al dataset
columns_to_check_outliers = ['velocidad_registrada']
df_original = remove_outliers_iqr(df_original.copy(),columns_to_check_outliers)
````

## Ejercicio 04

> Puntos: 2

Definir una clase FineAnalyzer que cumpla con las siguientes funcionalidades:

- Reciba el DataFrame limpio como argumento de inicialización.

- Encapsule correctamente los datos.

- Implemente métodos para:

  - Ranking de las patentes más multadas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de patentes y cantidad..

  - Ranking de los horarios donde se producen más multas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de hora y cantidad.


  - Exceso promedio de velocidad. Debe retorna un flotante.

  - Exceso real promedio de velocidad . Debe retorna un flotante.

  - Contabilizar la cantidad de multas por cada  ubicación. Debe retornar un DataFrame ordenado por ubicación mostrando la ubicación y la cantidad.

- Crear el objecto e invocar cada uno de los métodos en celdas separadas.
  

````python
class FineAnalyzer:
    """Clase para analizar datos de multas de tránsito."""

    def __init__(self, df: pd.DataFrame) -> None:
        """
        Inicializa el analizador con una copia del DataFrame original.

        Args:
            df: DataFrame con la información de las multas.
        """
        self._df = df.copy()

    def ranking_patentes(self) -> pd.DataFrame:
        """
        Obtiene las 5 patentes con mayor cantidad de infracciones.

        Returns:
            DataFrame con el ranking de patentes y sus cantidades.
        """
        if "patente" not in self._df.columns:
            raise ValueError("La columna 'patente' no existe en el DataFrame.")

        ranking = (
            self._df["patente"]
            .value_counts()
            .head(5)
            .reset_index()
        )
        ranking.index = ranking.index + 1
        ranking.columns = ["patente", "cantidad"]
        return ranking

    def ranking_horarios(self) -> pd.DataFrame:
        """
        Identifica los 5 horarios con mayor frecuencia de multas.

        Returns:
            DataFrame con el ranking de horas y sus cantidades.
        """
        if "hora" not in self._df.columns:
            raise ValueError("La columna 'hora' no existe en el DataFrame.")

        ranking = (
            self._df["hora"]
            .value_counts()
            .head(5)
            .reset_index()
        )
        ranking.index = ranking.index + 1
        ranking.columns = ["hora", "cantidad"]
        return ranking

    def exceso_promedio(self) -> float:
        """
        Calcula el promedio general del exceso de velocidad.

        Returns:
            Valor decimal del promedio.
        """
        if "exceso_velocidad" not in self._df.columns:
            raise ValueError("Falta la columna 'exceso_velocidad'.")

        return float(self._df["exceso_velocidad"].mean())

    def exceso_real_promedio(self) -> float:
        """
        Calcula el promedio del exceso de velocidad real.

        Returns:
            Valor decimal del promedio real.
        """
        if "exceso_velocidad_real" not in self._df.columns:
            raise ValueError("Falta la columna 'exceso_velocidad_real'.")

        return float(self._df["exceso_velocidad_real"].mean())

    def multas_por_ubicacion(self) -> pd.DataFrame:
        """
        Contabiliza el total de multas agrupadas por ubicación.

        Returns:
            DataFrame con la ubicación y su respectiva cantidad.
        """
        if "ubicacion" not in self._df.columns:
            raise ValueError("La columna 'ubicacion' no existe en el DataFrame.")

        ubic = (
            self._df["ubicacion"]
            .value_counts()
            .reset_index()
        )
        ubic.columns = ["ubicacion", "cantidad"]
        return ubic
````
